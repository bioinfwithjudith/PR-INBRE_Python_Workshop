```python
sequence = 'ATGTGGGAGA'
if 'TAA' in sequence:
  print(f'TAA in {sequence}')
elif 'TAG' in sequence:
  print(f'TAG in {sequence}')
elif 'TGA' in sequence:
  print(f'TGA in {sequence}')
else:
  print(f'A stop codon does not exist in {sequence}')
```
