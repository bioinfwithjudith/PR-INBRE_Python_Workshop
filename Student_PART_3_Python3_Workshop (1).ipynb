###Lesson 3.2.3b: Create a Panda DataFrame from multiple Lists

```python
#Use this block of code for notes
crtam_human = ['CRTAM',393,'human']
crtam_mouse = ['CRTAM',393,'mouse']
cadm1_human = ['CADM1',442,'human']

pd.DataFrame([crtam_human,crtam_mouse,cadm1_human],
             columns=['gene','length','organism'])
```

```python
gene_dictionary = {'CRTAM_HUMAN':{'length':393,'organism':'human'},
                   'CRTAM_MOUSE':{'length':781,'organism':'mouse'},
                   'CADM1_HUMAN':{'length':442,'organism':'human'}}

df = pd.DataFrame(gene_dictionary)
df = df.T
df = df.reset_index()
df = df.rename(columns={'index':'gene'})
df

df2 = pd.DataFrame(gene_dictionary).T.reset_index().rename(columns={'index':'gene'})
df2
```

```python
df2.to_csv('gene_dictionary_with_tab.csv',sep='\t')
df2.to_csv('gene_dictionary_with_comma.csv',sep=',')
```

##for loop through the line of a file

```python
fasta_file = open('crtam.faa','r')
for fasta_line in fasta_file:
  print(fasta_line)
fasta_file.close()
```

##while loop through the line of a file

```python
fasta_file = open('crtam.faa','r')
fasta_lines = fasta_file.readline()
while fasta_lines:
  print(fasta_lines)
  fasta_lines = fasta_file.readline()
fasta_file.close()
```
